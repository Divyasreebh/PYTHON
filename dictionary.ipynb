```python
#DICTIONARY : key  (it can  not be duplicate )+ values(it can  be duplicate )

Books={"python":200,"sql":100,"Excel":900,"ml":200}

#Books[GenAi]=200 in dic its to add any item

Books["python"]=900 #IF THE KEY IS THERE THEN IT WILL UPDATE
print(Books)

for i in Books.keys():
  print(i)

for i in Books.values():
  print(i)

for i in Books.items():
  print(i)
```
