```python
#append- adding the no. at the last 
#insert- inserting the number in particular position
#extend- adding mutliple nos at the last 

L=[10,20,30,40,50,60]

L.append(80)

L.insert(2,90)

L.extend([1,2,3,4])

print(L)

```

```python
# ADD G AT LAST
# ADD Z AT 3 POSITION 
# THEN ADD THREE MORE NAME

users=["A","B","C","D"]
users.append("G")
users.insert(3,"Z")
users.extend(["honey","divya","kiran"])

users[2]="DS" #updating the particular position given


print(users)

```

```python
#pop is removing, or removing by postion

#remove´it removes bx particular number we give

L=[10,20,30,40,50,60]

L.append(80)
L.insert(2,90)
L.extend([1,2,3,4])


print(L)

L.pop()
L.pop(3) #removes particular postion starts wiht 0

L.remove(80) # directly remobes the number

print(L)
```

```python
users=["A","B","C","D","A","B","C","A","B","C"]

print(users.count("B")) #counts - how many B are present
print(users.index("C")) #index - shows which posotion the element is present

users.sort() #sort- sorting in ascending order
users.sort(reverse=True) # reverse- gives in descending order

print(users) 
```
