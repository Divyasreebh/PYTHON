```python
EDA (Exploratory Data Analysis)
1. Full Form of EDA
EDA stands for Exploratory Data Analysis.

Definition
Exploratory Data Analysis (EDA) is the process of examining, understanding, summarizing, and visualizing a dataset to identify patterns, relationships, missing values, and unusual data before applying machine learning algorithms.

Why do we perform EDA?
EDA helps us:

Understand the dataset.
Identify missing values.
Find duplicate records.
Detect incorrect or inconsistent data.
Identify outliers.
Understand the distribution of data.
Find relationships between variables.
Select suitable features for machine learning.
2. Steps of EDA
Step	Name	Purpose
1	Load Dataset	Import the dataset into Python.
2	Understand Data	Check rows, columns, and data types.
3	Check Missing Values	Identify empty or null values.
4	Check Duplicates	Find repeated records.
5	Check Inconsistent Data	Identify incorrect values or inconsistent categories.
6	Statistical Summary	Calculate mean, median, minimum, maximum, etc.
7	Univariate Analysis	Analyze one variable at a time.
8	Bivariate Analysis	Analyze the relationship between two variables.
9	Multivariate Analysis	Analyze relationships among multiple variables.
10	Outlier Detection	Identify unusually high or low values.
11	Correlation Analysis	Measure relationships between numerical variables.
12	Data Cleaning	Correct data quality problems identified during analysis.
13	Feature Selection	Identify useful variables for machine learning.
Note: EDA and data cleaning are closely related. We often clean the data, perform further analysis, and repeat the process when necessary.

3. Simple Example
Suppose we have the following employee dataset:

Employee	Age	Salary	Experience
Amit	25	30,000	2
Priya	30	45,000	5
Rahul	28	40,000	4
Neha	35	60,000	10
Simran	200	50,000	6
By performing EDA, we may discover:

Missing values: Check whether any employee information is missing.
Duplicates: Check whether any employee record appears more than once.
Inconsistent data: Check whether age, salary, or experience contains invalid values.
Outliers: Simran's age of 200 is unusual and requires investigation.
Statistical analysis: Calculate the average salary and average age.
Relationship analysis: Investigate whether salary increases with experience.
EDA helps us understand these problems before building a machine learning model.

4. Types of EDA
A. Univariate Analysis
Definition: Analysis of a single variable.

Examples:

Distribution of employee ages.
Average employee salary.
Distribution of product prices.
Common techniques: Histogram, boxplot, and frequency counts.

B. Bivariate Analysis
Definition: Analysis of the relationship between two variables.

Examples:

Salary vs. Experience.
Sales vs. Advertising Cost.
Age vs. Income.
Common techniques: Scatter plot, bar plot, and correlation analysis.

C. Multivariate Analysis
Definition: Analysis of relationships among three or more variables.

Examples:

Salary, experience, age, and education.
Product price, discount, sales, and profit.
Common techniques: Heatmap, pair plot, and multiple-variable visualizations.

5. EDA Workflow in Machine Learning
The complete workflow is:

Load Data → Understand Data → Check Missing Values → Check Duplicates → Check Inconsistent Data → Statistical Summary → Univariate Analysis → Bivariate Analysis → Multivariate Analysis → Outlier Detection → Data Cleaning → Feature Selection → Model Building

```
