```python
import numpy as np

a=np.array([12,20,30,43,56,78])

print(a[a>=40])

print(np.sum(a))

print(np.max(a))

#same count, min, mean std median
```

```python
import numpy as np

marks=np.array([12,20,30,43,56,78])

print(marks+5)
print(marks.ndim)
print(marks.shape)
print(marks.size)
print(marks.dtype)
```
